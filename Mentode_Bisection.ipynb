# Persamaan Non Linear - Metode Bisection
Mencari akar f(x)=0 pada selang [a, b] dengan f(a)·f(b) < 0.
**Ide:** belah selang di tengah c=(a+b)/2, pilih setengah yang masih mengapit akar, ulangi sampai selang cukup kecil.

## Cell 1 - Import library
`numpy` untuk fungsi matematika, `matplotlib` untuk grafik, `pandas` untuk tabel iterasi.

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

## Cell 2 - Definisikan fungsi
Contoh: f(x) = x^3 - x - 2. Ganti sesuai soal.

```python
def f(x):
    return x**3 - x - 2
```

## Cell 3 - Fungsi bisection
Parameter: selang `a`, `b`, toleransi `tol` (lebar selang), batas iterasi `maks_iter`.
Syarat: f(a) dan f(b) berbeda tanda, jika tidak akar tidak dijamin ada.

```python
def bisection(f, a, b, tol=1e-6, maks_iter=100):
    if f(a) * f(b) > 0:
        raise ValueError("f(a) dan f(b) harus berbeda tanda")
    riwayat = []
    for i in range(1, maks_iter + 1):
        c = (a + b) / 2                       # titik tengah
        riwayat.append((i, a, b, c, f(c)))
        if f(c) == 0 or (b - a) / 2 < tol:    # akar tepat atau selang cukup kecil
            break
        if f(a) * f(c) < 0:                   # akar di [a, c]
            b = c
        else:                                 # akar di [c, b]
            a = c
    return c, riwayat
```

## Cell 4 - Jalankan
Tabel menunjukkan perubahan selang tiap iterasi.

```python
akar, riwayat = bisection(f, 1, 2)
tabel = pd.DataFrame(riwayat, columns=["iterasi", "a", "b", "c", "f(c)"])
print(tabel.to_string(index=False))
print("\nAkar hampiran:", akar, "| f(akar) =", f(akar))
```

## Cell 5 - Grafik
Grafik f(x) dengan titik akar hasil bisection.

```python
xs = np.linspace(0.5, 2.5, 200)
plt.plot(xs, f(xs), label="f(x)")
plt.axhline(0, color="gray")
plt.scatter([akar], [f(akar)], color="red", zorder=3, label="akar")
plt.legend(); plt.grid(True); plt.show()
```
